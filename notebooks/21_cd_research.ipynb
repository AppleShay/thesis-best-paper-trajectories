```python
import pandas as pd
import numpy as np
import requests
import time
from pathlib import Path

ROOT = Path("..")
MATCHED = ROOT / "data" / "matched"
FIGURES = ROOT / "data" / "figures"

df_cd = pd.read_csv(MATCHED / "cd_results.csv")
df_main = pd.read_csv(MATCHED / "huang_matched_openalex.csv")
df_main['work_id'] = df_main['openalex_id'].str.extract(r'(W\d+)')

high_cd = df_cd[df_cd['cd_5'] >= 0.8].merge(
    df_main[['work_id','paper_title','conference','year']],
    on='work_id', how='left'
)
print(f"Papers with CD ≥ 0.8: {len(high_cd)}")
print(high_cd[['paper_title','conference','year','cd_5','ni','nj']].to_string())
```

```python
print(df_cd.columns.tolist())
print(df_main.columns.tolist())
```

```python
high_cd = df_cd[df_cd['cd_5'] >= 0.8].merge(
    df_main[['work_id', 'paper_title']],
    on='work_id', how='left'
)
print(f"Papers with CD ≥ 0.8: {len(high_cd)}")
print(high_cd[['paper_title','conference','year','cd_5','ni','nj']].to_string())
```

```python
# Trust CD=1 only with ≥15 citers
reliable_high = high_cd[high_cd['ni'] + high_cd['nj'] >= 15]
print(reliable_high[['paper_title','conference','year','cd_5','ni']].to_string())
```

Papers with fewer than 10 citers within the 5-year window are excluded as CD scores are unreliable at low citation counts (Bornmann et al., 2020).
